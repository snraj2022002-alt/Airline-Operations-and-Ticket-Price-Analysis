```python
print(df['source_city'].mode())
print(df['departure_time'].mode())
print(df['stops'].mode())
print(df['flight'].mode())
print(df['arrival_time'].mode())
print(df['destination_city'].mode())
print(df['class'].mode())          #This column in categorical. so, Choose mode methord
```

```python
df['source_city'].fillna('Delhi',inplace=True)
df['departure_time'].fillna('Morning',inplace=True)
df['stops'].fillna('one',inplace=True)
df['flight'].fillna('0.00E+00',inplace=True)
df['arrival_time'].fillna('Night',inplace=True)
df['destination_city'].fillna('Mumbai',inplace=True)
df['class'].fillna('Economy',inplace=True)          #Column in fill the null data
```

```python
df['duration'].unique()           #Flight in traveling dertion find the outlar.so,use the median methord
```

```python
df['duration'].fillna(df['duration'].median(),inplace=True)
df['duration'].median()           #Outlear find column in fill the null data
```

```python
df['price'].unique()            #Flight in traveling dertion find the outlar.so,use the median methord
```

```python
df['price'].fillna(df['price'].median(),inplace=True)
df['price'].median()           #Outlear find column in fill the null data
```

```python
df.isnull().sum()             #This dataset in no null data
```

```python
df.drop('days_left',axis=1,inplace=True)
```

```python
df['price'] = pd.to_numeric(df['price']).astype('Int64')
df['price'].dtypes                #Coliumn price in float datatype chenge the integer datatype
```

```python
df=df.rename(columns={'index':'Index','airline':'Airlines','flight':'Flight_Way_Code','source_city':'Source_city','departure_time':'Starting_Time','stops':'Stops','arrival_time':'Arrivel_Time','destination_city':'Destination_city','class':'Class','duration':'Traveling_Duration','price':'Ticket_price'})
df                #datas in column name praperly changed the new name
```

```python
def Ticket_price_in_outliers(box):                 #stands for Interquartile Range
    Q1 = box['Ticket_price'].quantile(0.25)        #81382 records
    Q3 = box['Ticket_price'].quantile(0.75)        #240780 records
    IQR = Q3 - Q1                                  #Range 159398 records
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return box[(box['Ticket_price'] >= lower_bound) & (box['Ticket_price'] <= upper_bound)]
df= df.groupby('Class', group_keys=True).apply(Ticket_price_in_outliers)
```

```python
print(f"Original dataset rows: {len(df)}")
print(f"Cleaned dataset rows: {len(df)}")
print(f"Removed {len(df) - len(df)} outliers.")
```

```python
def analyze_airline(airline_name):
    Airline_column = df[df['Airlines'] == airline_name]

    if Airline_column.empty:
        print(f"No data found for Airline: {airline_name}")
        return

    print(f"Total Revenue: {Airline_column['Ticket_price'].sum():,}")
    print(f"Total Flights: {Airline_column['Index'].count():,}")
    print(f"Average Travel Duration: {Airline_column['Traveling_Duration'].mean():.2f} hours\n")

    print("--- Passenger Booking Class Counts ---")
    print(Airline_column['Class'].value_counts().to_string())

    print("--- Source City Counts ---")
    print(Airline_column['Source_city'].value_counts().to_string())

    print("--- Starting Time Counts ---")
    print(Airline_column['Starting_Time'].value_counts().to_string())

    print("--- Stop Counts ---")
    print(Airline_column['Stops'].value_counts().to_string())

    print("--- Top 5 Flight Codes ---")
    print(Airline_column['Flight_Way_Code'].value_counts().head(5).to_string())

    print("--- Arrival Time Counts ---")
    print(Airline_column['Arrivel_Time'].value_counts().to_string())

    print("--- Destination City Counts ---")
    print(Airline_column['Destination_city'].value_counts().to_string())
```

```python
for index in zip(range(1,6,1)):            #dataframe all airlines name form performance show function created show the all Airline in analysis
  Airline_name=df['Airlines'].unique()[index]
  print(f'======={index}====={Airline_name} in for Analysis')
  analyze_airline(Airline_name)
```

```python
plt.figure(figsize=(8, 4))                  #Airline wish total revanue

plt.plot(Area_Groupby.index, Area_Groupby.values, marker='o', linestyle='--', color='green')
plt.fill_between(Area_Groupby.index, Area_Groupby.values, color='lightgreen', alpha=0.3)

plt.title('Total Revenue by Airline')
plt.xlabel('Airlines')
plt.ylabel('Total Revenue (in Billions)')
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```
